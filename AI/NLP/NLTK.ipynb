# Stemming
Rmove suffix to get root form of word  ( running --> run)
- porterstemmer - suffix stripping
- lancasterstemmer - it not work for suffix, removesa at leat something
- snowballstemmer - we have to define language

```python
from nltk.stem import PorterStemmer
pst = PorterStemmer()
```

```python
pst.stem('affection')
```

```python
pst.stem('maximum')
```

```python
words_to_stem = ['give', 'giving','given','gave']

for words in words_to_stem:
    print(words + ':' + pst.stem(words))
```

```python
words_to_stem = ['give', 'giving','given','gave', 'thinking','loving','maximum']

for words in words_to_stem:
    print(words + ':' + pst.stem(words))
```

```python
from nltk.stem import LancasterStemmer
lst = LancasterStemmer()

for words in words_to_stem:
    print(words + ':' + lst.stem(words))
```

```python
from nltk.stem import SnowballStemmer
sbst = SnowballStemmer('english')

for words in words_to_stem:
    print(words + ':' + sbst.stem(words))
```

```python
print(SnowballStemmer.languages)
```

# POS(parts of speech)

```python
sent = 'Sam is a natural when it comes to the drawing'
sent_tokens = word_tokenize(sent)
sent_tokens
```

```python
# nltk.download('averaged_perceptron_tagger')
for token in sent_tokens:
    print(nltk.pos_tag([token]))
```

```python
sent2 = 'John is eating adelicious cake'
sent2_tokens = word_tokenize(sent2)

for token in sent2_tokens:
    print(nltk.pos_tag([token]))
```

# NER(Named Entity Recognition)

```python
#nltk.download('maxent_ne_chunker')
#nltk.download('words')
  
from nltk import ne_chunk
```

```python
NE_sent = 'The US president stays in WHITEHOUSE'
```

```python
NE_tokens = word_tokenize(NE_sent)
NE_tokens
```

```python
NE_tags = nltk.pos_tag(NE_tokens)
NE_tags
```

```python
NE_NER = ne_chunk(NE_tags)
print(NE_NER)
```
