```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv(r"E:\Vaishnavi\practiced\Datasets\titanic dataset.csv", header= 0, dtype={'Age':np.float64})
df
```

```python
df.describe()
```

```python
# Name column is not useful while deciding survival of a person, let's delete it
del df["Name"]
df.head()
```

```python
del df['Ticket']
df.head()
```

```python
del df['Cabin']
df.head()
```

```python
del df['Fare']
df.head()
```

```python
# changing values for male female values to numeric values ,male=1, female=2

def get_number(str):
    if str =='male':
        return 1
    else:
        return 2

# created new column Gender filled with 1 or 2 based on column Sex

df['Gender'] = df['Sex'].apply(get_number)
df.head()
```

```python
del df['Sex']
df.head()
```

```python
df.isnull().sum()
```

```python
mean = df[df['Survived']==1].Age.mean()
mean
```

```python
df['age'] = np.where(pd.isnull(df.Age) & df.Survived ==1 , mean, df.Age)
df.head(2)
```

```python
df.isnull().sum()
```

```python
# we already filled null for Survived==1
# now remaining all null will be for Survived ==0

meanS = df[df.Survived==0].Age.mean()
meanS
```

```python
df.age.fillna(meanS,inplace=True)
df.head()
```

```python
df.isnull().sum()
```

```python
del df['Age']
df
```

```python
# find the Survived people who are boarded from different port

survivedQ = df[(df['Embarked']=='Q') & (df['Survived']==1)].shape[0]
survivedC = df[(df['Embarked']=='C') & (df['Survived']==1)].shape[0]
survivedS = df[(df['Embarked']=='S') & (df['Survived']==1)].shape[0]
print(survivedQ)
print(survivedC)
print(survivedS)
```

```python
survivedQ = df[(df['Embarked']=='Q') & (df['Survived']==0)].shape[0]
survivedC = df[(df['Embarked']=='C') & (df['Survived']==0)].shape[0]
survivedS = df[(df['Embarked']=='S') & (df['Survived']==0)].shape[0]
print(survivedQ)
print(survivedC)
print(survivedS)
```

```python
df.dropna(inplace=True)
df.head()
```

```python
df.isnull().sum()
```

```python
df.shape
```

```python
# renaming column

df.rename(columns={'Gender':'Sex','age':'Age'},inplace=True)
df.head()
```

```python
def getEmb(str):
    if str == 'S':
        return 1
    elif str == 'Q':
        return 2
    else:
        return 3

df['Embark'] = df['Embarked'].apply(getEmb)
df.head()
```

```python
del df['Embarked']
df.rename(columns={'Embark':'Embarked'},inplace=True)
df.head()
```

```python
# Drawing a pie chart for number of males and females aboard
import matplotlib.pyplot as plt
from matplotlib import style

male = (df['Sex']==1).sum()         # calculating total number of males
female = (df['Sex']==2).sum() 
print(male)
print(female)

p = [male,female]

plt.pie(p,
        labels = ['male', 'female'],     # giving corresponding labels
        colors = ['green','yellow'],     # corresponding colors
        explode = (0.15,0),              # gap between pie slices
        startangle = 0,                  # start angle
        autopct = '%.2f%%')
plt.axis('equal')
plt.show()
```

```python
# more prcise pie chart

maleS = df[(df['Sex']==1) & (df['Survived']==1)].shape[0]
maleN = df[(df['Sex']==1) & (df['Survived']==0)].shape[0]
femaleS = df[(df['Sex']==2) & (df['Survived']==1)].shape[0]
femaleN = df[(df['Sex']==2) & (df['Survived']==0)].shape[0]
print(maleS)
print(maleN)
print(femaleS)
print(femaleS)
```

```python
chart=[maleS,maleN,femaleS,femaleN]
colors=['lightskyblue','yellowgreen','Yellow','Orange']
labels=["Survived Male","Not Survived Male","Survived Female","Not Survived Female"]
explode=[0,0.05,0,0.1]
plt.pie(chart,labels=labels,colors=colors,explode=explode,startangle=100,counterclock=False,autopct="%.2f%%")
plt.axis("equal")
plt.show()
```
