 - Function without argument - do not require anything
 - Function with argument devided into two part
 1. Formal Argument
 2. Actual Argument

```python
def add(x,y):      # x and y are --- FORMAL ARGUMENT
    c = x + y
    return c

add(4,5)           # 4 and 5 are --- ACTUAL ARGUMENT
```

Argument:
1. Positional Argument
2. Keyword Argument
3. Default Argument
4. Variable Argument

Positional Argument
- matches position with respective argument

```python
def add(x, y, z, m, n):
    c = x+y
    return c

add(4, 5, 0, 7, 4)       # x=4, y=5, z=0, m=7, n=4
```

```python
def person(name,age):
    print(name)
    print(age)

person('nit',22)          # name='nit',age=22
```

```python
def person(name,age):
    print(name)
    print(age)

person(22,'nit')          # name=22, age='nit'
```

```python
def person(name,age):
    print(name)
    print(age + 1)        # string can only be concatenate with another string and not a integer

person(22,'nit')          # name=22, age='nit'
```

Keyword Argument
- need to write keyword for each and every argument

```python
def person(name,age):
    print(name)
    print(age + 1)

person(age = 22, name ='nit')          # age = 22, name ='nit'
```

```python
def person(name,age,salary):
    print(name)
    print(age + 1)
    print('got placed')

person(age = 22, name ='nit')     #  Formal Argument = Actual Argument; else give error
```

```python
def person(name,age,age2):
    print(name)
    print(age)
    print(age2)

person(age = 20, name ='nit', age2 = 21)    
```

Default Argument
- if given use that value or else take default value  

```python
def person(name,age):
    print(name)
    print(age)

person('nit')        # requires two argument 
```

```python
def person(name,age=18):
    print(name)
    print(age)

person('nit')              # now by default it takes age = 18
```

```python
def person(name,age=18):
    print(name)
    print(age)

person('nit',age = 40)    # if given use that value oe else take default value  
```
