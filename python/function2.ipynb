# variable length
 *at last formal argument | other argument as positinal argument | last formal argument accept remaining all actual argument as tuple

```python
def sum(a,b):
    c = a+b
    return c
    
sum(5,6)
```

```python
def sum(a,b):
    c = a+b
    return c
    
sum(5,6,7,8,9)
```

```python
def sum(a,*b):
    c = a+b
    return c
    
sum(5,6,7,8,9)   # a=5 as a int ; b=(6,7,8,9)  as atuple
```

```python
def sum(a,*b):
    print(type(a))
    print(type(b))
    
sum(5,6,7,8,9)   # a=5 as a int ; b=(6,7,8,9)  as atuple
```

```python
def sum(a,*b):
    c = a
    for i in b :
        c = c + i
    return c
sum(5,6,7,8,9)   # a=5 as a int ; b=(6,7,8,9)  as atuple

```

# kwargs (keyworded variable length arguments)
** at last argument | other argument keyword argument | last argument accept variable length(remaining) arguments as a tuple

```python
def person(name,*data):
    print(name)
    print(data)
person('Alex',36,'JOHN',924385)
```

```python
def person(name,*data):
    print(name)
    print(data)
person('Alex',age=36,friend='JOHN',phone_no = 924385) 
# as we added keyword arguments it will give error. to solve add *
```

```python
def person(name,**data):
    print(name)
    print(data)
person('Alex',age=36,friend='JOHN',phone_no = 924385) 
# as we added keyword arguments it will give error. to solve add *
# this way we can give as many keyword arguments as we want
```

# global
The global keyword is used inside a function to indicate that a variable refers to a global variable, not a local one.

```python
x = 10     # global variable

def update_x():
    global x            # declare that we are using global variable x
    x+=5                # modify global variable

update_x()
print(x)
```

```python
# if want to define/modify global variable inside function

a = 10

def something():
    global a
    b = 15
    print('in function',b) # local
    print('global variable',a)

something()
print('out function',a)    # gloabl
```

```python
a = 20

def something():
    global a      # refer global a | nnow we can modify it inside function
    a = 15
    print('in function',a) # local
    a = 9
something()
print('out function',a)    # gloabl

```

# globals
- globals() is a built-in function that returns a dictionary representing the current global symbol table. It allows you to access and modify global variables programmatically.

```python
 x= 10 # global variable

def update_x():
    globals()['x'] += 5   # access and modify global variable using dictionary
update_x()
print(x)   
```

```python
# if we used local & global in the same function this is not good idea thats wy introduced to GLOBALS

a = 10
print(id(a))         # id/address where that variable value stored | address is given to '10' not an 'a'

def something():
    a = 9                   # local | has nothing to do with global a
    x = globals()['a']
    print(id(x))
    print('in function',a)  # local

something()
print('out function',a)
```

```python
# if we used local & global in the same function this is not good idea thats wy introduced to GLOBALS

a = 10
print(id(a))         

def something():
    a = 9                   
    x = globals()           # if i don't mention a then it will create new memory
    print(id(x))
    print('in function',a)  
    
    globals()['a'] = 15     # here we change the value of global a 

something()
print('out function',a)
```

# Recurssion

```python
def wish():
    print('Hello')
    print('Hi')

wish()
```

```python
def wish():
    print('Hello')
    print('Hi')
    wish()

wish()

# after a certain numeber of execution (till reaches recursion limit) it gives error: max recursion depth exceeded
# this max recursion limit is different for different envirnments
```

# How to use lambda with other function
- filter(): apply filter to get value that meet certain criteria
- map(): apply a certain operation/function on each element of an object(ex.list)
- reduce():

```python
# find list of even numbers
def is_even(n):
    return n%2 == 0               # return element if condition True
    
num = [1,2,3,4,5,6]
even = list(filter(is_even,num))   # num elements are used as arguments for is_even fuction
even

# each element of list num are passed to the function is_even one by one
```

```python
# find list of odd numbers
def is_odd(n):
    return n%2 != 0               # return element if condition True
    
num = [1,2,3,4,5,6]
odd = list(filter(is_odd,num))   # num elements are used as arguments for is_even fuction
odd
```

```python
# find even numbers using lambda
# lambda helps for code optimization

num = [1,2,3,4,5,6]

even = list(filter(lambda x: x%2 == 0 , num))   
odd = list(filter(lambda x: x%2 != 0 , num))  

print('odd:',odd)
print('even:',even)
```

```python
# map(): apply operation to all element of list

num = [1,2,3,4,5,6]

even = list(filter(lambda x: x%2 == 0 , num))    # filter outs elements that meets certain condition
double = list(map(lambda x: x*2 , even))         # apply the operation on each element of that list

print('even:',even)
print('double:',double)              
```

```python
# reduce()

from functools import reduce

def add_all(a,b):
    return a+b

num = [1,2,3,4,5,6]

even = list(filter(lambda x: x%2 == 0 , num))    # filter outs elements that meets certain condition
double = list(map(lambda x: x*2 , even))         # apply the operation on each element of that list

sums = reduce(add_all, double)
sums
    
```

```python
# filter(), map(), reduce()

num = [1,2,3,4,5,6]

even = list(filter(lambda x: x%2 == 0 , num))    # filter outs elements that meets certain condition
double = list(map(lambda x: x*2 , even))         # apply the operation on each element of that list
sums = reduce(lambda a,b : a+b, double)          # reduces whole object(list) to a single value using given operation

print('even:',even)
print('double:',double)              
print('sums:',sums)              
```
